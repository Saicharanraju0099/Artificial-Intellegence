### Cosine similarity check
**In vector databases, cosine similarity is used to compare embeddings (numerical representations of data) to find items with similar meanings, regardless of differences in magnitude.**

### Euclidean distance
**In a vector database context:Every piece of data (a document, an image, a product, etc.) is stored as an embedding vector.When you query with another vector, the database calculates the Euclidean distance between your query vector and each stored vector.The smaller the distance, the closer (more similar) the items are.**

```python
from sklearn.metrics.pairwise import euclidean_distances
```

```python
euclidean_distances([query_embedding], document_embeddings)             # here lowest value is the most similar one
```

| **Metric**                       | **Similarity / Distance Score Range**                                | **What It Measures**                                     | **Behavior in Vector DB**                                                                                                                         |
| -------------------------------- | -------------------------------------------------------------------- | -------------------------------------------------------- | ------------------------------------------------------------------------------------------------------------------------------------------------- |
| **Cosine Similarity**            | -1 to 1 (often 0 to 1 in practice since embeddings are non-negative) | Cosine of the angle between two vectors (direction only) | Ignores vector length; two vectors with same “vibe” but different magnitudes are still close. Great for semantic similarity in text, images, etc. |
| **Euclidean Distance**           | 0 to ∞                                                               | Straight-line (L2 norm) distance between two points      | Sensitive to magnitude; larger vectors are “further” even if pointing in same direction. Useful for geometric or clustering tasks.                |
| **Dot Product**                  | -∞ to ∞                                                              | Raw alignment of two vectors (magnitude × direction)     | Bigger values if vectors are both aligned **and** large. Good for ranking when vector norms encode importance (like in transformers).             |
| **Manhattan Distance** (L1 norm) | 0 to ∞                                                               | Sum of absolute differences across dimensions            | More robust to outliers than Euclidean. Looks like a grid-based “city block” distance. Sometimes used when sparsity matters.                      |

| Index Type                                          | How It Works                                                                                          | Speed                                     | Memory Usage                                           | Accuracy                                  | Best For                                                                           |
| --------------------------------------------------- | ----------------------------------------------------------------------------------------------------- | ----------------------------------------- | ------------------------------------------------------ | ----------------------------------------- | ---------------------------------------------------------------------------------- |
| **Flat (Brute-force)**                              | Stores all vectors as-is. Every query checks distance to *every* vector.                              | Slow for large datasets (linear search).  | High (stores full dataset).                            | 100% exact.                               | Small datasets or when you need *exact* results.                                   |
| **IVF (Inverted File Index)**                       | Splits space into clusters (via k-means). Query compares only inside nearest clusters.                | Much faster than Flat for large datasets. | Lower than Flat (stores centroids + compressed codes). | Approximate, depends on #clusters probed. | Huge datasets where you can trade a little accuracy for speed.                     |
| **HNSW (Hierarchical Navigable Small World Graph)** | Builds a graph where vectors are connected to neighbors; queries “walk” the graph to find candidates. | Very fast (logarithmic-like).             | Higher (extra graph connections stored).               | High, close to exact.                     | Large-scale search with high recall and speed needs (like recommendation systems). |


| Dataset Size (# vectors) | Recommended Index                                              | Why                                                                           |
| ------------------------ | -------------------------------------------------------------- | ----------------------------------------------------------------------------- |
| **Up to \~100k**         | **Flat (IndexFlatL2 or FlatIP)**                               | Memory is manageable, brute-force gives 100% accuracy and is simple.          |
| **\~100k – 1M**          | **HNSW**                                                       | Much faster than Flat, very high recall, good balance of speed and accuracy.  |
| **1M – 10M**             | **IVF (e.g., IVF1000)**                                        | Clustering reduces search space dramatically, keeps latency reasonable.       |
| **10M – 100M+**          | **IVF with PQ (Product Quantization)** or **IVF+HNSW hybrid**  | IVF narrows candidates, PQ compresses memory, HNSW speeds up within clusters. |
| **100M – 1B+**           | **IVF+PQ+OPQ (Optimized PQ)**, possibly **disk-based indexes** | At this scale, memory is the bottleneck, so quantization is needed.           |

```python
pages = []
async for page in loader.alazy_load():
    pages.append(page)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
splited_docs=splitter.split_documents(pages)  # it will split the documents into smaller chunks based on the chunk size and overlap specified
splited_docs
```

**we need to have context(retriver),prompt(hub),model(google),parser(langchain) for chaining in generation process**

```python
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import  RunnablePassthrough
```

### Chaining

```python
def format_docs(docs):
    return "\n".join([f"Document {i+1}: {doc.page_content}" for i, doc in enumerate(docs)])
```

```python
rag_chain = (
    {"context": second_retriever | format_docs,"question":RunnablePassthrough() }
    | prompt
    | model
    | StrOutputParser()

)
```

```python
rag_chain.invoke("what is big data analytics?")
```
