```python
from langchain_huggingface import HuggingFaceEmbeddings
embeddings = HuggingFaceEmbeddings(model="all-MiniLM-L6-v2")
```

```python
embeddings.embed_query("Hello Langchain")
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
documents = ['what is the capital of India',
             'what is the capital of USA',
             'who is the prime minister of India',]
```

```python
my_query = "Narendra modi is the prime minister of which country?"
```

```python
document_embeddings=embeddings.embed_documents(documents)
```

```python
document_embeddings
```

```python
query_embedding=embeddings.embed_query(my_query)
```

```python
query_embedding
```

### Cosine similarity check
**In vector databases, cosine similarity is used to compare embeddings (numerical representations of data) to find items with similar meanings, regardless of differences in magnitude.**


```python
cosine_similarity([query_embedding], document_embeddings)      # here highest value is the most similar one
```

### Euclidean distance
**In a vector database context:Every piece of data (a document, an image, a product, etc.) is stored as an embedding vector.When you query with another vector, the database calculates the Euclidean distance between your query vector and each stored vector.The smaller the distance, the closer (more similar) the items are.**

```python
from sklearn.metrics.pairwise import euclidean_distances
```

```python
euclidean_distances([query_embedding], document_embeddings)             # here lowest value is the most similar one
```

| **Metric**                       | **Similarity / Distance Score Range**                                | **What It Measures**                                     | **Behavior in Vector DB**                                                                                                                         |
| -------------------------------- | -------------------------------------------------------------------- | -------------------------------------------------------- | ------------------------------------------------------------------------------------------------------------------------------------------------- |
| **Cosine Similarity**            | -1 to 1 (often 0 to 1 in practice since embeddings are non-negative) | Cosine of the angle between two vectors (direction only) | Ignores vector length; two vectors with same “vibe” but different magnitudes are still close. Great for semantic similarity in text, images, etc. |
| **Euclidean Distance**           | 0 to ∞                                                               | Straight-line (L2 norm) distance between two points      | Sensitive to magnitude; larger vectors are “further” even if pointing in same direction. Useful for geometric or clustering tasks.                |
| **Dot Product**                  | -∞ to ∞                                                              | Raw alignment of two vectors (magnitude × direction)     | Bigger values if vectors are both aligned **and** large. Good for ranking when vector norms encode importance (like in transformers).             |
| **Manhattan Distance** (L1 norm) | 0 to ∞                                                               | Sum of absolute differences across dimensions            | More robust to outliers than Euclidean. Looks like a grid-based “city block” distance. Sometimes used when sparsity matters.                      |


```python
import faiss
from langchain.vectorstores import FAISS
from langchain_community.docstore.in_memory import InMemoryDocstore
```

```python
index=faiss.IndexFlatL2(384)      # 384 is the dimension of the embedding model
index
```

```python
vector_store=FAISS(
    embedding_function=embeddings,
    index=index,
    docstore = InMemoryDocstore(),
    index_to_docstore_id={},
)
```

```python
vector_store.add_texts(["AI is powerful", "AI is the future", "AI is the new source of Knowledge"], ids=["1", "2", "3"])
```

```python
vector_store.index_to_docstore_id
```

```python
result=vector_store.similarity_search("AI is the new source of Knowledge", k=2)
```
