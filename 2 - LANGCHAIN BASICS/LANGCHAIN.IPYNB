### Prompt Engineering
**Prompt engineering is the practice of designing and refining the input (the “prompt”) given to an AI model to optimize its output for accuracy, relevance, and usefulness.**

```python
from langchain_core.prompts import ChatPromptTemplate
prompt = ChatPromptTemplate.from_messages(
    [
        ("system","You are a helpful assistant provide me answers based on the questions asked"),
        ("user","{input}")
    ]
)
prompt
```

```python
from langchain_groq import ChatGroq
model = ChatGroq(model="openai/gpt-oss-20b")
print(model)
```

### Combining the prompt template with the model

```python
# Chaining
chain = prompt  | model
chain
```

```python
response=chain.invoke({"input":"What is OpenAI? say something about it in ten lines"})
print(response.content)
```
