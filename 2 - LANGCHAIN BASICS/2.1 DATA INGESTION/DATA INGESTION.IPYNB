### Data Ingestion
**Data ingestion is the process of collecting raw data from various sources and moving it into a system where it can be stored, processed, and analyzed. It’s basically the first step in any data pipeline—taking in information from places like databases, APIs, files, or sensors, and preparing it so other tools (like AI models or analytics systems) can actually use it.**

```python
# LOADING A PDF FILE
from langchain_community.document_loaders import PyPDFLoader
loader = PyPDFLoader('./data/Researchpaper.pdf')
docs = loader.load()
docs  # EVERY SINGLE PAGE IS CONVERTED TO ONE SEPERATED DOCUMENT
```

```python
## WEBBASE LOADER
from langchain_community.document_loaders import WebBaseLoader
import bs4
loader = WebBaseLoader(web_path=('https://medium.com/@tahirbalarabe2/%EF%B8%8Flangchain-vs-langgraph-a-comparative-analysis-ce7749a80d9c'))
docs = loader.load()
docs
```

```python
from langchain_community.document_loaders import WebBaseLoader
import bs4
loader = WebBaseLoader(web_path=('https://lilianweng.github.io/posts/2023-06-23-agent/'),
                       bs_kwargs=dict(parse_only = bs4.SoupStrainer(
                           class_ = ("post-title","post-content","post-header")
                       )))
docs = loader.load()
docs
```
