### HUGGING FACE VECTOR EMBEDDINGS

```python
os.environ["HF_TOKEN"] = os.getenv("HF_TOKEN")
```

Sentence Transformers are pretrained models (on Hugging Face) that turn sentences or paragraphs into dense vector embeddings that capture meaning.

Built on BERT/transformer models, but fine-tuned so similar sentences map to nearby vectors.

Used for semantic search, clustering, similarity, and retrieval.

Popular models: all-MiniLM-L6-v2, multi-qa-mpnet-base-dot-v1.

Work offline, free to use, and easy to load with the sentence-transformers Python library.

```python
from langchain_huggingface import HuggingFaceEmbeddings
embeddings = HuggingFaceEmbeddings(model="all-MiniLM-L6-v2")
```

```python
embeddings
```

```python
text = "this is hugging face sentence transformers"
query_result = embeddings.embed_query(text)
query_result
```
