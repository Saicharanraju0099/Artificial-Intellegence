### Tavily search to crawl the internet

**Tavily is a search API designed for AI agents, mainly used to give them real-time, relevant web information. Unlike Google or Bing APIs that return raw links, Tavily provides clean, summarized snippets that LLMs can directly use. Its main purpose is to fill the knowledge gap of AI models by letting them fetch up-to-date facts, news, or context, which makes it especially useful in agents and retrieval-augmented generation (RAG) systems.**

```python
from langchain_community.tools.tavily_search import TavilySearchResults
```

```python
import os
from dotenv import load_dotenv
load_dotenv()

```

```python
import os

os.environ["TAVILY_API_KEY"] = os.getenv("TAVILY_API_KEY")

```

```python
tool=TavilySearchResults()
```

```python
tool.invoke("what are the latest advancements in artificial intelligence?")
```
